```python
import pandas as pd
import numpy as np

import random
```

```python
# set random seed for reproductivity
random.seed(42)
```

```python
# Function to generate random date and time within range
def generate_random_datetime(start_date, end_date):
    start_timestamp = pd.to_datetime(start_date).timestamp()
    end_timestamp = pd.to_datetime(end_date).timestamp()
    random_timestamp = random.uniform(start_timestamp, end_timestamp)
    return pd.to_datetime(random_timestamp, unit = 's')
```

```python
# Function to generate random product name
def generate_product_name():
    products = ['T-shirts', 'Jeans', 'Sneakers', 'Smartphone', 'Headphones', 'Watch', 'Backpack','Laptop']
    return random.choice(products)
```

```python
# Function to generate random price
def generate_price():
    return round(random.uniform(10,500), 2)
```

```python
# Function to generate random quantity
def generate_quantity():
    return random.randint(1,10)
```

```python
# Function to generate random Customer name
def generate_customer_name():
    first_names = [
    "John", "Emma", "Michael", "Sophia", "David",
    "Olivia", "James", "Ava", "William", "Mia"
    ]

    last_names = [
        "Smith", "Johnson", "Brown", "Davis", "Wilson",
        "Miller", "Taylor", "Anderson", "Thomas", "Moore"
    ]
    return random.choice(first_names)+' '+ random.choice(last_names)
```

```python
# Function to generate random email
def generate_email():
    domains = ["gmail.com", "yahoo.com", "outlook.com", "hotmail.com"]
    return f"{random.randint(1,999)}@{random.choice(domains)}"
```

```python


cities = [
    "Delhi", "Mumbai", "Bengaluru", "Kolkata",
    "Chennai", "Hyderabad", "Pune", "Jaipur"
]

states = [
    "Delhi", "Maharashtra", "Karnataka", "West Bengal",
    "Tamil Nadu", "Telangana", "Rajasthan"
]

def generate_address():
    house_no = random.randint(1, 9999)
    city = random.choice(cities)
    state = random.choice(states)
    pincode = random.randint(100000, 999999)

    return f"{house_no}, {city}, {state} - {pincode}"
```

```python
# Numbers of row in dataFrame
num_rows = 1000
```

```python
# Generate data for Data Frame
data = {
'OrderID': ['ORD' + str(i).zfill(4) for i in range(1, num_rows + 1)],
'DateTime': [generate_random_datetime('2022-01-01', '2022-12-31') for _ in range(num_rows)],
'Product': [generate_product_name() for _ in range(num_rows)],
'Price': [generate_price() for _ in range(num_rows)],
'Quantity': [generate_quantity() for _ in range(num_rows)]
}
```

```python
# Calculate TotalPrice
data['TotalPrice'] = [round(data['Price'] [i] * data['Quantity'][i], 2) for i in range(num_rows)]

# Generate customer-related data
data['CustomerName'] = [generate_customer_name() for _ in range(num_rows)]
data['Email'] = [generate_email() for _ in range(num_rows)]
data['Address'] = [generate_address() for _ in range(num_rows)]

#Create DataFrame
ecommerce_df = pd. DataFrame(data)

# Display the DataFrame
ecommerce_df.head()
```

```python
#Load the data into a pandas dataFrame named ecommerce_df
#ecommerce_df = pd.read_csv('sales_data.csv')

#Convert the 'DateTime' column to datetime format
ecommerce_df['DateTime'] = pd.to_datetime(ecommerce_df ['DateTime'])

#Assuming you want to analyze sates for a specific time period (start_date and end_date)
start_date= '2022-01-01' # Replace this with your desired start date
end_date ='2022-03-31'    #Replace this with your desired end date

#Filter the DataFrame for the specified time period
sales_period = ecommerce_df[(ecommerce_df['DateTime'] >= start_date) & (ecommerce_df['DateTime'] <= end_date)]

#Calculate total sales revenue by multiplying 'Price' and 'Quantity', then summing up

total_revenue = (sales_period['Price']* sales_period['Quantity']).sum()

print("Total sales revenue for the period from", start_date, "to", end_date, "is:", total_revenue)
```

```python
import pandas as pd

#Assuming ecommerce_df contains your DataFrame with the given columns

#Group by Product and calculate total sales
total_sales_by_product = ecommerce_df.groupby('Product') ['TotalPrice'].sum().reset_index()

#Sort the results by TotalPrice in descending order
total_sales_by_product = total_sales_by_product.sort_values(by= 'TotalPrice', ascending=False)

#Display the result
print(total_sales_by_product)
```
