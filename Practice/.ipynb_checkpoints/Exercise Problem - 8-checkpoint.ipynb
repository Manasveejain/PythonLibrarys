## Exercise 8 — College Data Set

This exercise relates to the College data set, which can be found in the file `College.csv` on the book website. It contains a number of variables for 777 different universities and colleges in the US.

The variables are:

- **Private** : Public/private indicator
- **Apps** : Number of applications received
- **Accept** : Number of applicants accepted
- **Enroll** : Number of new students enrolled
- **Top10perc** : New students from top 10 % of high school class
- **Top25perc** : New students from top 25 % of high school class
- **F.Undergrad** : Number of full-time undergraduates
- **P.Undergrad** : Number of part-time undergraduates
- **Outstate** : Out-of-state tuition
- **Room.Board** : Room and board costs
- **Books** : Estimated book costs
- **Personal** : Estimated personal spending
- **PhD** : Percent of faculty with Ph.D.s
- **Terminal** : Percent of faculty with terminal degree
- **S.F.Ratio** : Student/faculty ratio
- **perc.alumni** : Percent of alumni who donate
- **Expend** : Instructional expenditure per student
- **Grad.Rate** : Graduation rate

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
college= pd.read_csv("csvfiles/college.csv")
college.head()
```

```python
college.shape
```

```python
college2 = pd.read_csv('csvfiles/College.csv', index_col=0)
college3 = college.rename({'Unnamed: 0': 'College'},
axis=1)
college3 = college3.set_index('College')
```

```python
college.describe()
```

```python
from pandas.plotting import scatter_matrix

features = ["Top10perc", "Apps", "Enroll"]
scatter_matrix(college[features])
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x = "Private" , y ='Outstate', data = college, palette = 'pastel')
plt.title('Outstate vs Private')
plt.show()
```

```python
college['Elite'].value_counts()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x = "Elite" , y ='Outstate', data = college, palette = 'pastel')
plt.title('Outstate vs Private')
plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.histplot(college['Books'], bins = 80, kde = True , color = 'purple')
plt.title('Distribution of Books')
plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.histplot(college['Apps'], bins = 15, kde = True , color = 'purple')
plt.title('Distribution of Apps')
plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.histplot(college['Grad.Rate'], bins = 20, kde = True , color = 'purple')
plt.title('Distribution of Grad.Rate')
plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.histplot(college['Expend'], bins = 30, kde = True , color = 'blue')
plt.title('Distribution of Expend')
plt.show()
```

### Brief summary of what I discovered

The notebook analyzes the **College dataset**, which contains information on **777 U.S. colleges and universities**.

Key findings from the analysis:

* The dataset contains variables related to **applications, enrollment, tuition, expenditure, graduation rate, student demographics, and college type**.
* **Private vs. public colleges:** The boxplot suggests that private colleges generally have **higher out-of-state tuition (Outstate)** than public colleges.
* A new qualitative variable, **`Elite`**, was created by classifying colleges based on whether **`Top10perc > 50`**.
* The **Elite vs. Outstate** boxplot indicates that elite colleges tend to have **higher out-of-state tuition** than non-elite colleges.
* A scatter-matrix was used to examine relationships among **`Top10perc`, `Apps`, and `Enroll`**, showing relationships between academic selectivity and application/enrollment numbers.
* Histograms were created for **Books, Applications (`Apps`), Graduation Rate (`Grad.Rate`), and Expenditure (`Expend`)** to understand their distributions.
* Overall, the analysis suggests that **college type and academic selectivity are associated with differences in tuition**, while application, enrollment, expenditure, and graduation variables show varying distributions across institutions.
